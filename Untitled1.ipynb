```python
import pandas as pd
import networkx as nx
import json
from itertools import combinations
from networkx.readwrite import json_graph

df = pd.read_csv("data_scopus.csv")
required_columns = ["Year", "Authors", "Authors with affiliations"]

# Treat blank or whitespace-only cells as missing.
df[required_columns] = df[required_columns].replace(
    r"^\s*$", pd.NA, regex=True
)

df = df.dropna(subset=required_columns)
df = df.dropna(subset=[
    "Year",
    "Authors",
    "Author(s) ID",
    "Authors with affiliations"
])

df = df.drop_duplicates(subset="EID")

G = nx.Graph()
skipped_rows = []
skipped_entries = []

for index, row in df.iterrows():
    author_ids = [
        item.strip()
        for item in str(row["Author(s) ID"]).split(";")
        if item.strip()
    ]

    author_entries = [
        item.strip()
        for item in row["Authors with affiliations"].split(";")
        if item.strip()
    ]

    if len(author_ids) != len(author_entries):
        skipped_rows.append(index)
        continue

    valid_authors = []

    for author_id, entry in zip(author_ids, author_entries):
        parts = [part.strip() for part in entry.split(",")]

        if len(parts) < 4:
            skipped_entries.append((index, entry))
            continue

        name = f"{parts[0]}, {parts[1]}"
        affiliation_start = 2

        if parts[2] in {"Jr.", "Jr", "Sr.", "Sr", "II", "III", "IV"}:
            name += f", {parts[2]}"
            affiliation_start = 3

        affiliation = ", ".join(parts[affiliation_start:-1])
        country = parts[-1]

        if not affiliation or not country:
            skipped_entries.append((index, entry))
            continue

        if author_id not in G:
            G.add_node(
                author_id,
                name=name,
                affiliation=affiliation,
                country=country
            )

        valid_authors.append(author_id)

    unique_authors = list(dict.fromkeys(valid_authors))

    for author1, author2 in combinations(unique_authors, 2):
        if G.has_edge(author1, author2):
            G[author1][author2]["value"] += 1
        else:
            G.add_edge(author1, author2, value=1)

# Export after processing ALL publications.
data = json_graph.node_link_data(G)

if "edges" in data:
    data["links"] = data.pop("edges")

with open("authors_network.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"Publications processed: {len(df)}")
print(f"Authors: {G.number_of_nodes()}")
print(f"Coauthor links: {G.number_of_edges()}")
print(f"Publications skipped for mismatched counts: {len(skipped_rows)}")
print(f"Incomplete author entries skipped: {len(skipped_entries)}")

print("\nFirst five skipped entries:")
for index, entry in skipped_entries[:5]:
    print(f"Row {index}: {entry!r}")
```
